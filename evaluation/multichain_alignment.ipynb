```python
import os
import numpy as np
import torch

from itertools import permutations

from Bio.PDB.MMCIFParser import MMCIFParser

cif_pred = '/cpfs01/projects-HDD/cfff-6f3a36a0cd1e_HDD/public/protein/workspace/wangjun/alphafold3-pytorch/evaluation/8sza_predict.cif'
cif_gt = '/cpfs01/projects-HDD/cfff-6f3a36a0cd1e_HDD/public/protein/workspace/wangjun/alphafold3-pytorch/evaluation/8sza-assembly1.cif'
```

```python
parser = MMCIFParser(auth_chains=False,auth_residues=False)
```

```python
def load_cif(path,return_tensor=True):
    name = os.path.basename(path)
    structure = parser.get_structure(name,path)
    model = structure[0] # 只存在一个model
    
    atom_coords = []
    atom_indices = []
    chain_auth_id_map = {}
    for idx,chain in enumerate(model.get_chains()):
        chain_auth_id_map[chain.id] = idx
        chain_atoms_coorda = [a.coord for a in chain.get_atoms()]
        atom_coords.extend(chain_atoms_coorda)
        atom_indices.extend([idx]*len(chain_atoms_coorda))

    atom_coords = np.array(atom_coords)
    atom_indices = np.array(atom_indices)
    r_ans = {
        'chain_auth_id_map':chain_auth_id_map,
        'atom_coords': torch.tensor(atom_coords) if return_tensor else atom_coords,
        'atom_indices':torch.tensor(atom_indices) if return_tensor else atom_indices
    }

    return r_ans
```

```python
import tree
from typing import List

pred = load_cif(cif_pred)
gt = load_cif(cif_gt)
```
